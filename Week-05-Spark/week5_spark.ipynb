```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.read.csv(
    "file:/Workspace/Users/srithankodumuru22@gmail.com/Drafts/Online Sales Data.csv",
    header=True,
    inferSchema=True
)
```

```python

df.show(5)
df.printSchema()
```

```python
df = df.withColumnRenamed("Transaction ID", "tid") \
.withColumnRenamed("Date", "dt") \
.withColumnRenamed("Product Category", "cat") \
.withColumnRenamed("Product Name", "prod") \
.withColumnRenamed("Units Sold", "qty") \
.withColumnRenamed("Unit Price", "price") \
.withColumnRenamed("Total Revenue", "rev") \
.withColumnRenamed("Region", "reg") \
.withColumnRenamed("Payment Method", "pay")
```

```python
df1 = df.dropDuplicates(["tid", "dt"])
df1.show()
```

```python
df.filter(col("reg") == "Europe").groupBy("cat").agg(avg("rev")).show()
```

```python
df.groupBy("reg") \
.count() \
.filter(col("count") > 100) \
.show()
```

```python
from pyspark.sql.functions import *

df = df.withColumn("age", lit(25))
df = df.withColumn("subscription", lit("Premium"))
df.filter((col("age") >= 18) & (col("age") <= 30) & (col("subscription") == "Premium")).show()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df = df.withColumn("raw_timestamp", col("dt"))
df = df.withColumn("raw_timestamp", col("raw_timestamp").cast(TimestampType()))
df = df.withColumnRenamed("raw_timestamp", "event_time")
df.select("dt", "event_time").show()
```

```python
from pyspark.sql.functions import *

df = df.withColumn("email", lit("user@gmail.com"))
df = df.withColumn("username", lit("user"))
df = df.filter(col("email").isNotNull() & (col("username") != ""))
df.select("email", "username").show()
```

```python
from pyspark.sql.functions import *

df.agg( min("price").alias("Min Price"),max("price").alias("Max Price"), avg("price").alias("Avg Price")
).show()
```

```python
df = df.withColumn("store_id", lit(1))
df = df.dropDuplicates()
df = df.na.fill(0, ["price"])
df.groupBy("store_id").sum("rev").show()
```
